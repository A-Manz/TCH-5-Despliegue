```python
import requests

datos = {
    "metros": 90,
    "habitaciones_limpio": 3,
    "baños_limpio": 2,
    "zona": "centro",
    "barrio": "Sol",
    "tipo_inmueble": "Piso",
    "planta_limpio": "3ª",
    "ascensor_limpio": "S",
    "localizacion_limpio": "EXTERIOR"
}

respuesta = requests.post("https://tch-5-despliegue.onrender.com/predict", json=datos)
respuesta.json()
```

## VALIDACIONES:

```python
datos_mal = dict(datos)
datos_mal["metros"] = -50

respuesta = requests.post("https://tch-5-despliegue.onrender.com/predict", json=datos_mal)
print(respuesta.status_code)
respuesta.json()
```

```python
datos_mal = dict(datos)
datos_mal["barrio"] = "Cierto ático famoso"

respuesta = requests.post("https://tch-5-despliegue.onrender.com/predict", json=datos_mal)
print(respuesta.status_code)
respuesta.json()
```
